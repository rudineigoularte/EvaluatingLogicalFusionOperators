# Função para realizar o histograma médio

```python
def AverageHistogram(histogram):
    #Copiando os histogramas do vídeo
    list = histogram.copy()

    #Inicializando variáveis
    AverageHist = []
    SumHist = [100]

    #Quantidade de histogramas do vídeo
    qtdHist = len(list)
    print(qtdHist)


    #Ciando um histograma com a soma de todos os histogramas no vídeo
    for hist in range (qtdHist):
      SumHist += list[hist]

    #Dividindo a soma pelo número de histogramas, para obter a média
    AverageHist = SumHist/qtdHist

    return(AverageHist)
```
