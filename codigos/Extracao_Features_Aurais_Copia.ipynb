#<h1>Função para Extração das Features Aurais</h1>

```python
## ==========
# Nova função para extrair features MFCC
## ==========

def extractMfcc(videoPath, shots, stream=0):
    video = VideoFileClip(videoPath)
    fps = video.fps
    audio = AudioFileClip(videoPath)
    featuresPerShot = []

    for i in range(shots.shape[0]):
        inicio = shots.iloc[i, 0] / fps
        fim = shots.iloc[i, 1] / fps

        audioShot = audio.subclip(inicio, fim).to_soundarray()
        if len(audioShot.shape) > 1:
            audioShot = audioShot[:,stream]
        # nfft is the greatest power of 2 closest to samplerate*winlen
        featuresPerShot.append(mfcc(audioShot, samplerate=44100, winlen=0.030, winstep=0.020, nfft=2048))

    audio.close()
    video.close()

    return featuresPerShot
```

#Script para extrair características aurais

```python

#Preparação dos dados
video = 'bbc_01.mp4'
shotspath = '/content/drive/MyDrive/IC_Leticia/DatasetBCC/GT/'+'bbc_01_shots.csv'
videopath = '/content/drive/MyDrive/IC_Leticia/DatasetBCC/movies/'+video

shots = pd.read_csv(shotspath, sep = ',', header = 0)

#Extração de características MFCC
auralFeatures = extractMfcc(videopath,shots)

#print("Tamanho auralFeatures: ",len(auralFeatures))
#print(len(auralFeatures[0][0]),auralFeatures[0][0])

#Gravando as featrures em arquivo
#A pasta "Features" deve existir dentro da Pasta de Trabalho"
with open('/content/drive/MyDrive/IC_Leticia/DatasetBCC/Features/aural_features_'+video+'.arq', 'wb') as arq:
    pickle.dump(auralFeatures, arq)
```
