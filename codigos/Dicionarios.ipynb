#<h1>Função para a geração de dicionários</h1>

```python
## ==========
# Os dados ("features" - vetores de características SIFT/MFCC/...)
# são colocados em um único espaço de características - uma bag (Bag of Features
# - BoF), representada pela variável "data". E são então clusterizados.
# Os centróides de cada agrupamento/cluster são chamados de palavras visuais/aurais/...
# e seu conjunto compõe uma "Bag of Visual/Aural/... Words (BoVW)" ou "Diconário de
# Palavras Visuais/Aurais/..., retornado na variável "centroids".
#
# O algoritmo de agrupamento utilizado é o K-Means. O hiperparâmetro k do
# K-Means é passado como argumento ("n") para função e vai determinar o tamanho
# do dicionário a ser montado.
#
# O valor ideal para k depende dos dados e pode ser obtido por métodos como
# Silhueta ou então Cotovelo. 
## ==========
def getBagOfFeatures(features, n):
    data = []
    for b in features: # cada b é uma coleção de vetores (SIFT ou MFCC) dentro de uma posição de features
      #print(b)
      if not(b is None):
        for v in b: #cada v é um vetor de características
          data.append(v)

    data = np.float32(np.asarray(data))
    criteria = (cv.TERM_CRITERIA_EPS + cv.TERM_CRITERIA_MAX_ITER, 100, 1.0)
    ret,labels,centroids = cv.kmeans(data, n, None, criteria, 10, cv.KMEANS_RANDOM_CENTERS)
    return centroids
```

#Script para a geração do dicionário de palavras visuais

```python
#Carcaterísticas visuais previamente extraídas e salvas em arquivo
with open('/content/drive/MyDrive/IC_Leticia/DatasetBCC/Features/visual_features_bbc_11.mp4.arq', 'rb') as arq_visual_features:
    visualFeatures = pickle.load(arq_visual_features)

#Geração do dicionário:
#O tamanho do dicionário é determinado pelo tamanho do agrupamento, 2o parâmetro
#da função.
visualDic = getBagOfFeatures(visualFeatures, 100)
print("Tamanho Dicionário Visual: ", len(visualDic))

#Salvando o dicionário em arquivo
#A pasta "dicionarios2" deve existir dentro de "MBA_Curso13_Dados"
with open('/content/drive/MyDrive/IC_Leticia/DatasetBCC/Dicionarios/visual_dic_bcc_11.arq', 'wb') as arq_visual_dic:
    pickle.dump(visualDic, arq_visual_dic)
```

#Script para a geração do dicionário de palavras aurais

```python
#Carcaterísticas aurais previamente extraídas e salvas em arquivo
with open('/content/drive/MyDrive/IC_Leticia/DatasetBCC/Features/aural_features_bbc_11.mp4.arq', 'rb') as arq_aural_features:
    auralFeatures = pickle.load(arq_aural_features)

#Geração do dicionário:
#O tamanho do dicionário é determinado pelo tamanho do agrupamento, 2o parâmetro
#da função.
auralDic = getBagOfFeatures(auralFeatures, 100)
print("Tamanho Dicionário Aural: ", len(auralDic))

#Salvando  o dicionário em arquivo
#A pasta "dicionarios2" deve existir dentro de "MBA_Curso13_Dados"
with open('/content/drive/MyDrive/IC_Leticia/DatasetBCC/Dicionarios/aural_dic_bcc_11.arq', 'wb') as arq_aural_dic:
    pickle.dump(auralDic, arq_aural_dic)

```
